# Introduction to APIs and Requests Module in Python 

- APIs allow different software to talk and share data  
- Python’s requests module makes using APIs easy  
- Today’s focus: Understanding APIs and how to send requests in Python

## What is an API?

- Stands for Application Programming Interface  
- A set of rules for software to communicate  
- Allows programs to send requests and get responses

**Example:** Weather app fetching current temperature

## Real-Life Analogy: Restaurant

**You** = Customer who wants food (client program)  
**Waiter** = API, takes your order and brings food  
**Kitchen** = Server, prepares the order (data/service)

- *The API connects client requests to server responses*

## How Does an API Work?

**Request**: You ask for something (e.g., data)  
**Processing**: The server processes your request  
**Response**: Server sends back the result or an error

## Types of APIs

**REST APIs**: Use HTTP methods (GET, POST, etc.), most popular  
**SOAP APIs**: Older style, uses XML data  
**GraphQL APIs**: Lets you request exactly what you want

## What is an API Endpoint?

- An endpoint is a URL where you send your API requests  
- Each endpoint provides specific data or services

**Example:** 

- `https://api.weather.com/forecast for weather info  `
- `https://api.twitter.com/tweets for tweets`

### Basic Example: Sending a GET Request

To start, let's send a basic `GET` request to fetch data from an API.


````python
import requests

# Define the URL of the API endpoint
url = "https://jsonplaceholder.typicode.com/posts"

# Make a GET request to the API endpoint
response = requests.get(url)

# Print the status code and content
print(f"Status Code: {response.status_code}")
print(f"Response Content: {response.json()}")
````

In this code:
- **GET request**: Requests data from the server.
- **Response**: Contains data sent by the server. Here, we are printing the JSON response which contains a list of posts.

### Example: Sending Data with POST
Now, let's send data using a POST request. The POST method is typically used when we need to send data to a server, such as creating a new resource.

````python
import requests

# API URL for POST request
url = "https://jsonplaceholder.typicode.com/posts"

# Data to send with the request
data = {
    "title": "New Post",
    "body": "This is the content of the new post.",
    "userId": 1
}

# Send the POST request
response = requests.post(url, json=data)

# Print status code and the response content
print(f"Status Code: {response.status_code}")
print(f"Response Content: {response.json()}")

````

In this code:
- `requests.post(url, json=data)`: Sends a POST request to the specified URL with the provided data.  
- `response.json()`: Returns the server's response after posting the data.

### Handling Errors
It's always good practice to handle errors in your requests, especially with dealing with multiple services.  
You can check if the request was successful using the status code.

````python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)

if response.status_code == 200:
    print("Request was successful.")
    print(f"Response Content: {response.json()}")
else:
    print(f"Error {response.status_code}: Unable to fetch data.")
````

````python
# 3️⃣ Handle error when fetching a non-existent post (404)

import requests

url = f"https://jsonplaceholder.typicode.com/99"

response = requests.get(url)

if response.status_code == 200:
    print("✔ Post found.")
    print(f"Response: {response.json()}")
elif response.status_code == 404:
    print(f"❌404 Error: Post not found.")
    print("The requested resource does not exist.")
else:
    print(f"❌Unexpected error: {response.status_code}")
    print(f"Response: {response.text}")
````

````python
import requests

# 4️⃣ Add query parameter to GET request (e.g. userId=1)

# Define the base URL of the API endpoint
url = "https://jsonplaceholder.typicode.com/posts"

# Define the query parameters as a dictionary
# The keys are the parameter names and the values are their corresponding values
params = {
    "userId": 1,
    "limit": 5  # Example of another query parameter
}

# Make the GET request with the query parameters
# The 'params' argument in requests.get() automatically handles URL encoding
response = requests.get(url, params=params)

# Print the full URL that was requested, including the query parameters
print(f"Requested URL: {response.url}")

# Print the status code of the response
print(f"Status Code: {response.status_code}")

# Print the JSON response content (if applicable)
if response.status_code == 200:
    print("Response JSON:")
    print(response.json())
else:
    print(f"Error: {response.text}")
````

````python
import requests

# 4️⃣ Add query parameter to GET request (e.g. userId=1)

def query_req(userId):
    # API endpoint url - CORRECTED
    url = "https://jsonplaceholder.typicode.com/posts"
    
    # CORRECTED: Use params dictionary for query parameters
    params = {'userId': userId}
    
    response = requests.get(url, params=params)  # CORRECTED: pass params

    if response.status_code == 200:
        print(f"Data fetched successfully.")
        posts = response.json()
        print(f"Number of posts found: {len(posts)}")
        
        # Show first few results
        for i, post in enumerate(posts[:3], 1):
            print(f"Post {i}: {post['title']}")
            
    elif response.status_code == 404:
        print(f"Error 404: Data not found.")
    else:
        print(f"❌ Unexpected error: {response.status_code}")
        print(f"Response: {response.text}")
    
    return response.json()

# Test the function
query_req(5)
````

````python
# 5️⃣ Send GET request with custom headers (like User-Agent)

import requests

# Define the URL you want to send the GET request to
url = "http://httpbin.org/headers"  # This URL echoes back the received headers

# Create a dictionary for your custom headers
# The keys are the header names (e.g., 'User-Agent', 'Accept-Language')
# The values are the desired header values
custom_headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/111.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.google.com/"
}

# Send the GET request with the custom headers
response = requests.get(url, headers=custom_headers)

# Print the response status code and the response text (which will contain the echoed headers)
print(f"Status Code: {response.status_code}")
print("Response Text (Echoed Headers):")
print(response.text)
````
