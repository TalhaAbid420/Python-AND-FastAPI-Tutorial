# **Operators in Python**

Python provides various operators to perform computations and logical operations. This notebook will cover different types of operators and their usage with examples.

## **Boolean Values**
A boolean data type represents two values: `True` or `False`. Booleans are commonly used in conditional expressions and comparisons.

```python
print(True)
print(False)
```

## **Assignment Operators**
Assignment operators are used to assign values to variables. For example, `=` assigns a value, while `+=` adds and assigns.

```python
# Assignment Operator Examples
x = 5       # Assign 5 to x
x += 10     # Equivalent to x = x + 10


print(f"X: {x}")
```

## **Arithmetic Operators**
Arithmetic operators are used for mathematical operations like addition, subtraction, multiplication, and division.

```python
# Arithmetic Operations
a, b = 10, 3
print('Addition:', a + b)
print('Subtraction:', a - b)
print('Multiplication:', a * b)
print('Division:', a / b)
print('Modulus:', a % b)
print('Floor Division:', a // b)
print('Exponentiation:', a ** b)
```

# Combining both Arithmetic and Compuond Operators

```python
x = 2
print(f"X: {x}")

x += 8
print(f"X after addition: {x}")

x -= 5
print(f"X after subtraction: {x}")

x *= 8
print(f"X after multiplication: {x}")

x /= 2
print(f"X after division: {x}")

x %= 7
print(f"X after modulus: {x}")

x //= 2
print(f"X after floor division: {x}")

x **= 8
print(f"X after exponentiation: {x}")
```

## **Comparison Operators**
Comparison operators compare values and return `True` or `False`.

```python
# Comparison Operators
x, y = 5, 10

print(f"x > y: {x > y}")
print(f"x < y: {x < y}")
print(f"x == y: {x == y}")
print(f"x != y: {x != y}")
print(f"x >= y: {x >= y}")
print(f"x <= y: {x <= y}")
```

## **Logical Operators**
Logical operators (`and`, `or`, `not`) are used to combine multiple conditions.

```python
# Logical Operators

a, b, c = True, False, True

print(f"a and b: {a and b}")
print(f"a or b: {a or b}")
print(f"not a: {not a}")
print(f"not b: {not b}")


print(f"\na and b and c: {a and b and c}")
print(f"a or b or c: {a or b or c}")
print(f"not a: {not a}")
print(f"not b: {not b}")
```
