# What are Methods In OOP?
- Methods are functions defined inside a class 
- They represent actions or behaviors objects can perform   
- Help organize code by bundling behavior with data   
- Types of methods:   
  - Instance Methods   
  - Class Methods   
  - Static Methods


## 🧰 Instance Methods = Methods bound to an Instance
Instance methods are functions defined insided a class that operate on an instance of the class. They take `self` as their first parameter, which refers to the specific instance calling the method. These methods can access and modify the attributes of the instance  and call other instance methods.

```python
class House:
    # Constructor to initialize the house with a specific color
    def __init__(self, color):
        self.color = color      # Assign the color to the house instance

    # Instance method to display the color of the house
    def show_color(self):
        print(f"This house is painted {self.color}.")   # Access and print the house's color

# Create an instance of the House class with the color 'red'
my_house = House('red')

# Call the instance method to show the color of the house
my_house.show_color()
```

```python
# Create a class Circle with radius attribute
# Add method to calculate area using the radius
# Understand how methods use instance data

class Circle:
    
    def __init__(self, radius):
        self.radius = radius    # Assign the radius to the circle instance
    
    def area_of_circle(self):
        import math
        return math.pi * (self.radius ** 2)
    
my_circle = Circle(5)       # Creating an instance of Circle with radius 5

print(f"My circle radius = {my_circle.radius}")

print(f"Area of my circle is {my_circle.area_of_circle():.2f}")

```

## 📐 Class Methods = Bound to the Class, Not the Object
A **class method** is a method that is bound to the class itself, not any one instance of the class. It can access or modify class-level state, which is shared among all instances. Unlike instance methods that receive `self`, class methods receive `cls` as the first argument--referring to class, not an individual object.

To define a class method, you use the `@classmethod` decorator above the method definition. Class methods are commonly used for factory methods or when you want behaavior that affects the class as a whole rather than just one instance.

```python
class House:
    def __init__(self, color, age):
        self.color = color      # Instance attribute: specific color of the house
        self.age = age      # Instance attribute: how old the house is

    @classmethod
    def from_construction_year(cls, color, construction_year):
        # This class method creates a House object by calculating its age
        house_age = 2025 - construction_year
        return cls(color, house_age)
    
# Create a new house using the class mehtod as a factory
my_house = House.from_construction_year("White", 1976)

# Print the age of the house
print(f"My house is {my_house.color} and it is {my_house.age} years old.")
```

### **Mini-challenge: Implement a class method that keeps count of the created instances**

```python
class Book:
    # Class attribute to keep track of number of Book instances
    count = 0

    def __init__(self, title):
        # Instance attribute to store the 'title' of the book
        self.title = title
        # Increment a class-level count whenever a new Book instance is created
        Book.count += 1

    @classmethod
    def get_count(cls):
        # Class method to return the total number of Book instances created
        return cls.count

# Create the first book instance
book1 = Book("The Great Gatsby")
# Create the second book instance
book2 = Book("To Kill a Mockingbird")
# Print the total number of Book instances created
print(f"Total number of Book instances created are: {Book.get_count()}")
print(f"Title of the books are '{book1.title}' and '{book2.title}'.")
```

```python
class Employee:
    company = "Tech Corp"      # class attribute
    employee_count = 0

    # Constructor to initialize name and salary
    def __init__(self, name, salary, company):
        self.name = name
        self.salary = salary
        self.company = company

        # Increment the employee count whenever a new Employee is created
        Employee.employee_count += 1

    
    @classmethod
    def change_company_name(cls, new_name):     # class method to change company name
        cls.company = new_name

    @classmethod
    def from_string(cls, emp_string):       # 
        """Create Employee from a string like 'Name-Salary-Company' """
        name, salary, company = emp_string.split('-')
        return cls(name, int(salary), company)
    
    @classmethod
    def get_employee_count(cls):
        return f"Total employee: {cls.employee_count}"

# Usage
# Create an employee using the regular constructor
emp1 = Employee("Alice", 70000, "Tech Corp")
print(f"Employee Name: {emp1.name}, Salary: {emp1.salary}, Company: {emp1.company}")
# Create an employee using the alternative constructor
emp2 = Employee.from_string("Bob-60000-Xeven Solutions")
print(f"Employee Name: {emp2.name}, Salary: {emp2.salary}, Company: {emp2.company}")
# Print total employee count
print(Employee.get_employee_count())
```

## Static Methods = Independent Utility Tools
A **static method** is a method that belongs to a class but doesn't access or modify the class or any of its instances. It doesn't take the `self`(the instance) or `cls`(the class) as its first argument. Instead, static methods behave like regular functions, but they're logically grouped with class because they perform tasks relevant to that class's purpose.

Static methods are ideal for utility operations -- validations, calculations, or helpers -- that do not depend on the instance's or class's internal state. You define them using the `@staticmethod` decorators.

```python
class House:
    def __init__(self, color, area):      # Constructor with color and area
        self.color = color
        self.area = area    # Area in square meters
    
    
    @staticmethod
    def is_valid_area(area):
        # Static method to validate if the area is reasonable
        return area > 0
    
# Use the static method to check if a given area is valid before creating a house
print(House.is_valid_area(-110))
print(House.is_valid_area(30))
print(House.is_valid_area(-58))
```

### **Mini-challenge: Write a static method to validate input before object creation**

```python
# Example combining static and class methods
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @staticmethod
    def is_valid_dimension(dimension):
        return dimension > 0
    
    @classmethod
    def create_rectangle(cls, width, height):
        if Rectangle.is_valid_dimension(width) and Rectangle.is_valid_dimension(height):
            return cls(width, height)
        else:
            raise ValueError("Width and Height must be positive.")
    
rect = Rectangle.create_rectangle(8, 6)
print(f"Rectangle = {rect.width}x{rect.height}")
```

```python
# Define a class to represent rectangular shapes
class Rectangle:
    # Constructor method - called when creating new instances
    def __init__(self, width, height):
        # Store width as an instance attribute (specific to each rectangle)
        self.width = width
        # Store height as an instance attribute (specific to each rectangle)
        self.height = height
    
    # Static method - doesn't need instance or class data, works like a utility function
    @staticmethod
    def is_valid_dimension(dimension):
        # Check if dimension is a number (int or float) AND positive
        return isinstance(dimension, (int, float)) and dimension > 0
    
    # Class method - operates on the class itself, used as alternative constructor
    @classmethod
    def create_rectangle(cls, width, height):
        # Validate both dimensions using the static method
        # cls refers to the Rectangle class (allows inheritance flexibility)
        if (cls.is_valid_dimension(width) and 
            cls.is_valid_dimension(height)):
            # If valid, create and return a new Rectangle instance
            # cls(width, height) calls the __init__ constructor
            return cls(width, height)
        else:
            # If invalid, raise an error to prevent creating bad objects
            raise ValueError("Width and Height must be positive numbers.")
    
    # Another class method - specialized constructor for squares
    @classmethod
    def create_square(cls, side):
        """Alternative constructor for squares"""
        # Validate the single side dimension
        if cls.is_valid_dimension(side):
            # Create square by using same value for width and height
            return cls(side, side)
        else:
            # Handle invalid side length
            raise ValueError("Side must be positive.")
    
    # Class method that parses string input - yet another constructor variant
    @classmethod
    def from_string(cls, dimensions_str):
        """Constructor from string like '10x20'"""
        # Try to parse the string - wrap in try/block for error handling
        try:
            # Split string by 'x' and convert both parts to floats
            # Example: "10x20" -> split -> ['10', '20'] -> map(float) -> [10.0, 20.0]
            width, height = map(float, dimensions_str.split('x'))
            # Use the existing create_rectangle method to validate and create
            # This reuses validation logic - don't repeat yourself!
            return cls.create_rectangle(width, height)
        except:
            # If anything goes wrong (parsing error, validation error, etc.)
            raise ValueError("Format must be 'widthxheight'")

# Demonstration of usage:

# Create rectangle using the main factory method
# Calls create_rectangle(8, 6) -> validates -> creates Rectangle(8, 6)
rect1 = Rectangle.create_rectangle(8, 6)    # Normal rectangle

# Create square using specialized constructor  
# Calls create_square(5) -> validates -> creates Rectangle(5, 5)
square = Rectangle.create_square(5)         # Square

# Create rectangle by parsing string input
# Calls from_string("10x15") -> parses -> creates Rectangle(10.0, 15.0)
rect2 = Rectangle.from_string("10x15")      # From string

# Display results using f-string formatting
# rect1.width = 8, rect1.height = 6
print(f"Rectangle: {rect1.width}x{rect1.height}")  # Output: Rectangle: 8x6

# square.width = 5, square.height = 5
print(f"Square: {square.width}x{square.height}")   # Output: Square: 5x5
```

## Comparison Table

# Python OOP Methods Comparison Table

| Feature | Instance Method | Class Method | Static Method |
|---------|-----------------|--------------|---------------|
| **Decorator** | None | `@classmethod` | `@staticmethod` |
| **First Parameter** | `self` (instance reference) | `cls` (class reference) | No special parameter |
| **Binding** | Bound to instance | Bound to class | Not bound |
| **Access to Instance Data** | Yes (via `self`) | No | No |
| **Access to Class Data** | Yes (via `self.__class__` or class name) | Yes (via `cls`) | No |
| **Can Modify Instance State** | Yes | No | No |
| **Can Modify Class State** | Yes | Yes | No |
| **Common Use Cases** | Operations on specific instances, data manipulation | Factory methods, alternative constructors, class-level operations | Utility functions, helper methods, validation |
| **Example** | `def get_area(self): return self.width * self.height` | `@classmethod def from_string(cls, text): return cls(*map(int, text.split()))` | `@staticmethod def validate_dimension(dim): return dim > 0` |
| **Called On** | Instance: `obj.method()` | Class or instance: `Class.method()` or `obj.method()` | Class or instance: `Class.method()` or `obj.method()` |
| **Inheritance Behavior** | Uses instance's actual class | Uses calling class (polymorphic) | Fixed implementation |

## Quick Summary

- **Instance Methods**: Work with object data - "I want to do something with THIS specific object"
- **Class Methods**: Work with class-level data - "I want to do something with the BLUEPRINT"  
- **Static Methods**: Utility functions - "I need a helper that's related to this class"

## Story-Driven Example (Combined)

```python
class MathTool():
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    # Instance method to calculate area
    def area(self):
        return MathTool.pi * (self.radius ** 2)
    
    # Class method to create a MathTool object from diameter
    @classmethod
    def from_diameter(cls, diameter):
        return cls(diameter / 2)
    
    # Static method to check if a value is a valid radius
    @staticmethod
    def is_valid_radius(radius):
        return radius > 0
    
# Example Usage
tool = MathTool(5)
print(tool.area())

tool_from_diameter = MathTool.from_diameter(10)
print(tool_from_diameter.radius)

print(MathTool.is_valid_radius(-1))
    
```

2️⃣ Create a class `Date` with attributes **year**, **month**, **day**. Implement a class method `from_string(date_string)` that takes a date string in the format "YYYY-MM-DD" and returns a `Date` object. Demonstrate its usage.

```python
class Date:
    def __init__(self, year, month, day):
        self.year = int(year)
        self.month = int(month)
        self.day = int(day)

    @classmethod
    def from_string(cls, date_string):
        """Create a Date object from a string in format 'YYYY-MM-DD'"""
        try:
            year, month, day = date_string.split('-')
            return cls(year, month, day)
        except ValueError:
            print("Error: Invalid format. Use 'YYYY-MM-DD'")
            return None

    def __str__(self):
        """Display date in readable format"""
        return f"{self.day:02d}/{self.month:02d}/{self.year}"

# Demonstrate usage
date1 = Date.from_string("2025-11-12")
print(f"Date created: {date1}")
print(f"Year: {date1.year}, Month: {date1.month}, Day: {date1.day}")

# Create another date
date2 = Date.from_string("1995-05-15")
print(f"Date created: {date2}")

# Test with invalid format
date3 = Date.from_string("12/11/2025")
```

3️⃣ Create a class `StringHelper` with a static method **is_palindrome** that checks if string is palindrome. Test it with atleast two examples.

```python
class StringHelper:
    
    @staticmethod
    def is_palindrome(text):
        """
        Check if a string is palindrome
        - Ignores case (uppercase/lowercase)
        - Ignores spaces
        - Only checks if it reads same forwards and backwards
        """
        if not isinstance(text, str):
            return False
            
        # Clean the text: lowercase and remove spaces
        cleaned = text.lower().replace(" ", "")
        
        # Check if it reads the same forwards and backwards
        return cleaned == cleaned[::-1]
    
    @staticmethod
    def explain_palindrome(text):
        """Show why a text is or isn't a palindrome"""
        cleaned = text.lower().replace(" ", "")
        is_pal = StringHelper.is_palindrome(text)
        
        print(f"Original: '{text}'")
        print(f"Cleaned: '{cleaned}'")
        print(f"Reversed: '{cleaned[::-1]}'")
        print(f"Palindrome: {is_pal}")
        print("-" * 30)

# Demonstration
print("=== PALINDROME EXPLANATION ===")

# Test examples
test_texts = [
    "racecar",
    "A man a plan a canal Panama", 
    "hello",
    "Madam",
    "noon",
    "python"
]

for text in test_texts:
    StringHelper.explain_palindrome(text)

# You can also use it directly
print("\n=== DIRECT USAGE ===")
print(f"Is 'racecar' palindrome? {StringHelper.is_palindrome('racecar')}")
print(f"Is 'hello' palindrome? {StringHelper.is_palindrome('hello')}")
```

4️⃣ Define a class `Toolbox` that has:
- An instance method `add_tool(tool_name)` that adds a tool to a list.
- A class method `get_total_toolboxes()` that returns how many **Toolboxes** has been created.
- A static method `is_valid_tool(tool_name)` that checks if the tool name is a normal name

```python
class Toolbox:
    total_toolboxes = 0  # Class variable to count toolboxes
    
    def __init__(self, owner):
        self.owner = owner
        self.tools = []  # Instance variable - each toolbox has its own tools
        Toolbox.total_toolboxes += 1
    
    # INSTANCE METHOD - works with specific toolbox
    def add_tool(self, tool_name):
        if Toolbox.is_valid_tool(tool_name):  # Using static method
            self.tools.append(tool_name)
            return f"Added '{tool_name}' to {self.owner}'s toolbox"
        else:
            return f"Invalid tool name: '{tool_name}'"
    
    def show_tools(self):
        return f"{self.owner}'s tools: {self.tools}"
    
    # CLASS METHOD - works with all toolboxes
    @classmethod
    def get_total_toolboxes(cls):
        return f"Total toolboxes created: {cls.total_toolboxes}"
    
    # STATIC METHOD - utility function (assuming "normal" = non-empty, only letters/spaces)
    @staticmethod
    def is_valid_tool(tool_name):
        """Check if tool name is valid (non-empty and only contains letters/spaces)"""
        if not isinstance(tool_name, str) or not tool_name.strip():
            return False
        # Check if contains only letters and spaces
        return all(char.isalpha() or char.isspace() for char in tool_name)

# Demonstration
print("=== TOOLBOX DEMONSTRATION ===")

# Create toolboxes
t1 = Toolbox("Alice")
t2 = Toolbox("Bob")

# Use instance method
print(t1.add_tool("hammer"))     # Valid
print(t1.add_tool("screwdriver")) # Valid  
print(t1.add_tool(""))           # Invalid - empty
print(t1.add_tool("drill123"))   # Invalid - has numbers
print(t1.add_tool("   "))        # Invalid - only spaces

print(t1.show_tools())

# Use class method
print(Toolbox.get_total_toolboxes())

# Use static method directly
print(f"Is 'wrench' valid? {Toolbox.is_valid_tool('wrench')}")
print(f"Is 'tool123' valid? {Toolbox.is_valid_tool('tool123')}")
```
