## **1. What is File I/O?**
File I/O (Input/Output) refers to how a program interacts with external files, such as **reading from** or **writing to** them. In Python, we use built-in functions like `open()` to work with files.

Examples of File I/O operations:
- Reading a text file.
- Writing data to a file.
- Working with structured data (JSON, CSV, etc.).


## **2. Why Do We Need File Handling?**
File handling is useful when we need to:
- Store and retrieve **persistent data**.
- Share data between different programs.
- Work with **large datasets** efficiently.

**Example Use Case:**
- Saving **user preferences** in an application.
- Storing **logs** for analysis.


## **3. Opening and Closing Files in Python**
To work with files in Python, we use the `open()` function.

**Syntax:**
```python
file = open('filename.txt', 'mode')
# Perform operations on the file
file.close()
```
Always close the file after opening it to free up system resources!


````python
# Opening a file and closing it properly
file = open('example.txt', 'w')
file.write("Hello, This is new Text!")
file.close()
````

##  **4. Reading from and Writing to Files**
We can use different modes while working with files:
- `'r'` → Read mode (default)
- `'w'` → Write mode (creates a new file or overwrites an existing one)
- `'a'` → Append mode (adds data without deleting existing content)
- `'r+'` → Read and write mode

 **Example:** Writing and reading a file in Python.


###  **Examples of Different File Modes in Python**
Below are examples demonstrating the different file modes available in Python:


####  **Read Mode (`'r'`)** - Opens a file for reading. The file must exist.

````python
try:
    with open('example.txt', 'r') as file:      # with this format, we don't need to close file.
        content = file.read()
        print(f"File content = {content}")
except FileNotFoundError as e:
    print(f"{e}")
````

####  **Write Mode (`'w'`)** - Creates a new file or overwrites an existing one.

````python
with open('example.txt', 'w') as file:
    file.write("This file is created by Talha Abid, for Talha Abid.")
print("Text written successfully!")
````

####  **Append Mode (`'a'`)** - Adds data to a file without deleting existing content.

````python
with open('example.txt','a') as file:
    file.write("This is a new line in a new file.\n")
print("Text appended successfully.")
````

####  **Read & Write Mode (`'r+'`)** - Opens the file for both reading and writing.

````python
with open('example.txt', 'r+') as file:
    current_content = file.read()
    print(f"Current Content: {current_content}")
    file.write("This is the last line of this file. Bye! Bye!\n")
    print("Text has read and written successfully!")
````

````python
# Writing to a file
with open('example.txt', 'w') as file:
    file.write("Negative thoughts destroy brain.")

with open('example.txt', 'r') as file:
    content = file.read()
    print(f"👉: {content}")
````

##  **5. Working with JSON Files in Python**
JSON (JavaScript Object Notation) is a popular format for **storing and exchanging data**.

Python provides a built-in `json` module to handle JSON files easily.

 **Example:** Storing and loading data using JSON files.


````python
import json

# Sample dictionary
data = {'name': 'Haris', 'age': 25, 'city': 'Lahore'}

# Writing JSON data to a file
with open('data.json', 'w') as file:
    json.dump(data, file)

# Reading JSON data from a file
with open('data.json', 'r') as file:
    loaded_data = json.load(file)
    print(loaded_data)
````

##  **6. Handling File Exceptions**
Sometimes, errors occur while working with files. For example:
- The file does not exist.
- The program lacks permission to read/write.
- The file is already in use by another program.

 **Use `try-except` to handle file-related exceptions.**


````python
try:
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print('Error: The file does not exist!')
````

##  **7. Best Practices in File Handling**
✔️ Always close the file after opening it (or use `with open()` for automatic closing).

✔️ Use `try-except` to handle file-related errors gracefully.

✔️ Avoid hardcoding file paths—use `os.path.join()` for compatibility.

✔️ Use `json.dump()` and `json.load()` for structured data instead of manually formatting text files.

````python
# 1️⃣Write a program that counts and prints the number of lines in a file.


try:
    with open('example.txt', 'r') as file:
        # count lines efficiently using a generator expression
        count = sum(1 for _ in file)
except FileNotFoundError:
    print("Error: 'example.txt' not found.")
else:       # it means: run this if try block succeded
    print("Number of lines:", count)
````

````python
# 2️⃣Write a program that asks the user for a filename and 
# checks whether it exists before opening it.

import os

filename = input("Enter the filename here: ")
if not filename:
    print("No filename provided.")
else:
    if os.path.exists(filename):
        try:
            with open(filename, 'r') as file:
                content = file.read()
                print("\nFile contents:")
                print(content)
        except PermissionError:
            print(f"Permission denied to read '{filename}'")
    else:
        print(f"File '{filename}' does not exist.")
````

````python
# 3️⃣Write a program that reads a file's content and writes a reversed version of it into a new file.

with open('example.txt', 'r') as file:
    content = file.read()
    reversed_content = content[::-1]

with open('reversed_example.txt', 'w') as file:
    file.write(reversed_content)
print("Reversed content written to 'reversed_example.txt' successfully.")
````

````python
# 4️⃣Write a program that copies the content of one file into another file.

with open('example.txt', 'r') as src_file:
    content = src_file.read()

with open('copied_file', 'w') as dest_file:
    dest_file.write(content)
    print(f"Text copied from '{src_file.name}' and pasted in '{dest_file.name}' successfully.")
````

````python
# 5️⃣Write a program that merges the content of two files into a single file.

try:
    with open('example.txt', 'r') as file1:
        content1 = file1.read()

    with open('reversed_example.txt', 'r') as file2:
        content2 = file2.read()

    with open('merged_file.txt', 'w') as merged_file:
        merged_file.write(content1 + '\n' + content2)

    print("Files merged successfully into 'merged_file.txt")

except FileNotFoundError as e:
    print(f"Error: One of the input files not found - {e}")
except PermissionError:
    print("Error: Permission denied accessing files")
except Exception as e:
    print(f"An unexpected error occured: {e}")
````

````python
# Create a logging system where every time the program runs, 
# it appends a new log entry with a timestamp into a log file.

from datetime import datetime
import time
import os

def log_event(message):
    """Append a timestamped log entry to the log file"""
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    log_entry = f"[{timestamp}] {message}\n"
    
    try:
        with open('program.log', 'a') as log_file:
            log_file.write(log_entry)
        print(f"Log entry added: {log_entry.strip()}")
    except PermissionError:
        print("Error: Permission denied to write to log file")
    except Exception as e:
        print(f"Error writing to log file: {e}")

# Example usage
log_event("Program started")

def unknown():
    time.sleep(1)  # Simulating work with sleep
    print("Doing some important work...")
unknown()

log_event("Operation completed successfully")
````
