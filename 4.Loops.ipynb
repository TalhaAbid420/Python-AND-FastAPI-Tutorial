## 1. Introduction to Loops

**What are Loops?**  
Loops are constructs in programming that allow you to execute a block of code repeatedly. They are essential for:
- Processing items in a collection (like a list or dictionary).
- Repeating an action until a certain condition is met.

**Why Use Loops?**  
- **Efficiency:** Automate repetitive tasks.
- **Simplicity:** Write less code to perform large-scale operations.
- **Flexibility:** Easily manage tasks such as iterating over data or implementing algorithms.



## 2. For Loops

The `for` loop in Python is used for iterating over a sequence (lists, strings, tuples, dictionaries, etc.). It processes each element one at a time in a very readable format.

**Basic Syntax:**

```python
for item in sequence:
    # process item


````python
# Example: Iterating over the list of numbers

numbers = [1, 2, 3, 4, 5]       # A simple list of numbers

for num in numbers:         # Loop through each number of the list
    print(num)          # Print the current number
````


### Using `enumerate()` in For Loops

Sometimes, you need both the item and its index. Python's built-in `enumerate()` function makes this easy.

**Teacher's Note:**  
Explain that `enumerate()` automatically adds a counter to the iterable, so you don't have to manage the index manually.

````python
# Example 2: Using enumerate to get both index and value from a list

Fruits = ["apples", "bananas", "cherries"]     # List of fruit names
for index, fruit in enumerate(Fruits):      # enumerate returns index and item
    print(f"Fruit-{index}: {fruit}")        # f-string used for clear output formatting
````

*Using for-loop without enumerate function:*

````python
# Initializing the counter variable
index = 1           

# Iterate over each fruit in the list
for fruit in Fruits:        
    print(f"Fruit-{index}: {fruit}")        # Print the current index  and the fruit
    index += 1                              # Manually increment the index counter, index = index + 1
````

### Iterating Over Dictionaries

Dictionaries are key-value pairs. You can iterate over:
- **Keys:** By default, looping over a dictionary returns its keys.
- **Values:** Using `.values()`
- **Key-Value Pairs:** Using `.items()`


````python
# Example 3: Iterating over a dictionary
person = {
    "name": "Bob",       # key: 'name', value: 'Bob'
    "age": 23,           # key: 'age', value: 23
    "city": "New York"   # key: 'city', value: 'New York'
    }

# Iterating over keys  (default behaviour)
print("Iterating over keys:\n")
for key in person.keys():
    print(f"{key}: {person[key]}")
````

````python
# Iterating over values

print("Iterating over values:\n")

for values in person.values():
    print(values)
````

````python
# Iterating over key-value pairs using .items()

print("Iterating over key-value pairs:\n")

for key, value in person.items():
    print(f"{key}: {value}")
````

## Iterating over list using for-loop
### same is with 'tuple' and 'set'

````python
a = [1, 3, 5, 7, 9]

for i in a:
    print(f"Value: {i}")


# We can use the range() method with for loop to traverse the list using index values.
print("\n Using range() to iterate:")
for i in range(len(a)):
    print(f"Index: {i}, Value: {a[i]}")

# We can also use the enumerate() function to iterate through the list. 
print("\n Using enumerate() to iterate:")
for index, value in enumerate(a):
    print(f"Index: {index}, Value: {value}")

# List comprehension: It is similar to for loop. 
# It provides the shortest syntax for looping through list.
print("\n Using list comprehension to iterate:")
[print(f"value: {value}") for value in a]

# Example: Squaring each value in the list using list comprehension
print("\n Squaring each value using list comprehension:")
squared = [value**2 for value in a]
print(f"Squared values: {squared}")
````

## 3. While Loops

**While Loops Overview:**  
The `while` loop runs as long as a specified condition is `True`.

**Basic Syntax:**

```python
while condition:
    # execute code block


````python
# Example: A basic while loop that prints numbers from 1 to 5

count = 1               # Initialize the counter variable   
while count <= 5:       # Loop will run as long as count is less than or equal to 5
    print(f"Count is: {count}")  # Print the current value of count
    count += 1          # Increment the counter variable by 1
````

## 4. Loop Control Statements

Loop control statements allow you to alter the normal flow of your loop.

1. **`break`:**  
   - Exits the loop immediately.
   - Useful when a certain condition is met and further processing is unnecessary.

2. **`continue`:**  
   - Skips the current iteration and moves to the next one.
   - Ideal for bypassing specific conditions without stopping the entire loop.

3. **`else`:**  
   - Runs when the loop completes naturally (i.e., it is not terminated by a `break` statement).
   - Can be used to execute code after the loop finishes processing.


````python
# Example 5: Using break, continue, and else in a loop

for num in range(11):     # loop over numbers from 0 to 10
    if num == 3:
        print("Skipping number 3")
        continue          # Skip the rest of the loop when num is 3
    if num == 8:
        print("Breaking the loop at number 8")
        break             # Exit the loop when num is 8
    print("Current number:", num)
else:
    # This else block runs if the loop completes without a break
    print("Loop completed without break.")
````

## 5. Advanced Looping Techniques

Beyond the basics, Python offers several advanced techniques to make your loops more concise and efficient.

### List Comprehensions

- **What are they?**  
  A compact way to process all or part of the elements in a collection and return a list with the results.

- **Syntax Example:**

```python
[expression for item in iterable if condition]


````python
# Example 6: List Comprehension to create a list of squares
squares = [x ** 2 for x in range(1, 11)]   # Compute the square for numbers 1 through 10
print("List of squares:", squares)
````

````python
# Example 7: Generator Expression to create cubes
cubes = (x ** 3 for x in range(1, 11))   # Create a generator for cubes of numbers 1 through 10
print("Cubes generated:")
for index,cube in enumerate(cubes, start=1):
    print(f"cube of {index}: {cube}")                       # Iterate through the generator to print each cube
    
````

````python
# Example 8: Using zip to iterate over two lists simultaneously
names = ['Alice', 'Bob', 'Charlie']   # List of names
ages = [25, 30, 35]                     # Corresponding list of ages
for name, age in zip(names, ages):      # zip pairs items from both lists together
    print(f"{name} is {age} years old.")
````

````python
# Example 9: Using itertools to cycle through a list
import itertools

colors = ['red', 'green', 'blue']   # List of colors
counter = 0                         # Counter to limit the loop
print("Cycling through colors using itertools.cycle:")
for color in itertools.cycle(colors):  # Cycle through colors indefinitely
    print(color)
    counter += 1
    if counter == 9:  # Stop after 6 iterations to avoid an infinite loop
        break
````

## Question 1: Basic Loop Patterns
Write a program that prints the following pattern using nested loops:

```python 
1
22
333
4444
55555
```

````python
for i in range(1, 6):
    for j in range(i):
        print(i, end='')
    print()
    

# Explanation:
# The outer loop (i) controls which number we're printing (1 through 5)
# The inner loop (j) controls how many times we print that number
# end="" prevents newlines after each print, and the empty print() creates the line break
# Alternative method uses string multiplication for cleaner code

for i in range(1, 6):
    print(str(i) * i)
print()
````

## Question 2: Data Processing with Loops

Given a list of numbers: *[12, 45, 23, 67, 34, 89, 15, 42]*

Use a loop to find and print all numbers greater than 30

Calculate and print the sum of all even numbers in the list

Count how many numbers are divisible by 3

````python
nums = [12, 45, 23, 67, 34, 89, 15, 42]

print("Numbers greater than 30:")
for num in nums:
    if num > 30:
        print(num)


even_sum = 0
for num in nums:
    if num % 2 == 0:
        even_sum += num
print(f"\nSum of even numbers is: {even_sum}")

nums_count = 0
for num in nums:
    if num % 3 == 0:
        nums_count += 1
print(f"\nCount of numbers divisible by 3 is: {nums_count}")
````

## Question 3: Advanced Loop Challenge
Write a program that:

Asks the user to input numbers until they enter "done"

Stores all valid numbers in a list

Uses loops to find:

The second largest number in the list

Numbers that are palindromes (same forwards and backwards)

````python
numbers = []
while True:
    user_input = input("Enter a number (or 'done' to finish):")
    if user_input.lower() == 'done':
        break
    try:
        numbers.append(int(user_input))
    except ValueError:
        print("Please enter a valid number.")

print(f"Your entered numbers are: {numbers}")


if len(numbers) < 2:
    print("Need at least two numbers for analysis")
else:
    unique_numbers = list(set(numbers))     # Remove duplicates
    unique_numbers.sort()
    second_largest = unique_numbers[-2] if len(unique_numbers) >= 2 else None
    print(f"Second largest number is: {second_largest}")


def is_palindrome(n):
    return str(n) == str(n)[::-1]

palindromes = [num for num in numbers if is_palindrome(num)]
print(f"Palindromic numbers are: {palindromes}")
````
