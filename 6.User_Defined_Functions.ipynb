## **Defining a Function in Python**
A function is a **block of reusable code** that performs a specific task. Instead of writing the same instructions repeatedly, we define a function once and call it whenever needed.

### **Key Components of a Function Definition**
1. **`def`** → A keyword that marks the beginning of the function.
2. **`function_name`** → A unique name for identifying the function.
3. **`parameters (optional)`** → Values passed to the function for processing.
4. **`:` (colon)** → Marks the end of the function header.
5. **Function body** → Contains the **logic (instructions)** that the function executes.
6. **`'''docstring'''`** → Used to describe what the function does.
7. **`return` (optional)** → Returns a result to the caller.


```python
# Defining a function named 'add' that takes two numbers as parameters
def add(num1, num2):
    """Funtion that adds two numbers and return the result"""
    print(f"Number 1 is: {num1}")       # printing num1
    print(f"Number 2 is: {num2}")       # printing num2
    add = num1 + num2                # adding both numbers with each other
    return add

# Calling the funciton and storing it in result variable
result = add(13, 26)
print(f"The sum of num1 and num2 = {result}")
```

Function can be declared without parameters.

```python
# Example: Define and call a greeting function.

# Define a function named 'greet' to print a welcome message
def greet():
    """Function that prints greetings"""
    print("Welcome to Python for Arificial Intelligence.")

# Calling the function
greet()
```

## **Function Arguments and Return Statement**
Function arguments allow us to **pass values** into a function, making it **more flexible and reusable**. There are **four types of arguments**:

1. **Default Arguments**: predefined values that can be overridden.
2. **Keyword Arguments**: passed by name to avoid order dependency.
3. **Required Arguments**: mandatory values passed when calling a function.
4. **Variable-Length Arguments**: accepting an arbitrary number of arguments.


```python
# Function with a default middle name
def full_name(first_name, last_name, middle_name="Abid"):
    """function that prints full name"""
    full_name = f"{first_name} {middle_name} {last_name}"
    print(full_name)

# Calling the funtion and passing first and last name
full_name("Talha", "Hussain")
```

```python
# Function with keyword arguments
def name(fname, mname, lname):          # fname, mname, lname are keywords also called parameters.
    """function that returns/print full name"""
    print(f"CONGRATULATIONS! {fname} {mname} {lname}")

# calling the function
name(mname="Abid", fname="Talha", lname="Hussain")
```

```python
# Required Arguments
# Function returning a full name
def name(fname, mname, lname):
    return "Hello, " + fname + " " + mname + " " + lname

# Calling the function and printing result
print(name("James", "Buchanan", "Barnes"))
```

```python
# Variable-length arguments
# Function with *args (Tuple of Arguments)
def Info(*args):
    """function that prints info from args"""
    for arg in args:
        print(arg)

Info("Talha", "Muslim", "Human")


# Another Example
def nums(*args):
    total = 0
    """function that totals nums and returns"""
    for arg in args:
        total += arg
        
    return f"\nTotal Numbers are: {total}"

print(nums(1, 5, 66, 18,2))
```

## Solution to Problem 3

```python
# Write a function convert_temperature that takes:  
# - temp_value temperature to convert 
# - unit (target unit: "C" for Celsius, "F" for Fahrenheit)  
# The function should convert and return the temperature based on the given unit.

def convert_temperature(temp_value, unit):
    """
    Converts temperature to the target unit: 'C' or 'F'. 
    If unit == 'C' treat temperature_value as Fahrenheit and convert to Celsius.
    If unit == 'F' treat temperature_value as Celsius and convert to Fahrenheit.
    """
    if unit == 'C'.strip():
        return (temp_value - 32) * 5/9
    elif unit == 'F'.strip():
        return (temp_value * 9/5) + 32
    else:
        raise ValueError("unit must be 'C' or 'F'.")
    
# Example usage
# print(convert_temperature(140, 'C'))

temperture = float(input('Enter temperature value: '))
unit = input("Enter the unit, you want to convert the temperature to: ").upper()

print(f"The temperature is {convert_temperature(temperture, unit)} {unit}.")
```

## Solution to Problem 4

```python
# Write a function `apply_discount` that takes:  
# - `original_price`
# - `discount_percentage`
# The function should **calculate the final price after discount** and return it.

def apply_discount(original_price, discount_percentage):
    if not (0 <= discount_percentage <= 100):
        raise ValueError("Discount percentage must be between 0 and 100.")
    price_after_discount = original_price - (original_price * discount_percentage / 100)
    return price_after_discount


print(apply_discount(100, 20))
print(apply_discount(1080, 80))
```

## Python Functions Practice Question: Inventory Management System
**Problem Statement**

You are building an inventory management system for a store. Create a program that:

Takes user input for multiple products with their details

Processes the data using functions

Generates reports with various analyses

**Requirements:**

Create the following functions:

`add_products()` - Takes user input for multiple products

`calculate_total_value(inventory)` - Calculates total inventory value

`find_most_expensive(inventory)` - Finds the most expensive product

`get_low_stock_items(inventory, threshold)` - Finds items with low stock

`generate_report(inventory)` - Generates a comprehensive report

**Input Format:**

For each product, user should enter:

- Product Name

- Price

- Quantity in Stock

The user should be able to enter multiple products until they type 'done'.

```python
def add_products():
    """
    Continuously takes user input for products until user types 'done'
    Returns a list of dictionaries containing product information
    """
    inventory = []
    while True:
        product_name = input("Enter product name (or 'done' to finish): ")
        if product_name.lower() == 'done':
            break
        print(f"\nProduct Name: {product_name}")
        try:
            product_price = float(input("Enter product price: "))
            print(f"Product Price: {product_price}")
            product_quantity = int(input("Enter product quantity: "))
            print(f"Product Quantity: {product_quantity}")
        except ValueError:
            print("Invalid input. Please enter numeric values for price and quantity.")
            continue

        product = {
            'name': product_name,
            'price': product_price,
            'quantity': product_quantity
        }
        inventory.append(product)

    return inventory

def calculate_total_value(inventory):
    """
    Calculates the total value of all products (price * quantity)
    """
    total_value = 0
    for item in inventory:
        total_value += item['price'] * item['quantity']
    return total_value

def find_most_expensive(inventory):
    """
    Returns the product with the highest price
    """
    if not inventory:
        return None
    most_expensive = max(inventory, key=lambda item: item['price'])
    return most_expensive

def get_low_stock_items(inventory, threshold=5):
    """
    Returns products with quantity less than threshold
    """
    low_stock = []
    # Your code here
    for item in inventory:
        if item['quantity'] < threshold:
            low_stock.append(item)
    return low_stock

def generate_report(inventory):
    """
    Generates a comprehensive inventory report
    """
    print("\n" + "="*50)
    print("          INVENTORY MANAGEMENT REPORT")
    print("="*50)
    
    # Your code here
    total_value = calculate_total_value(inventory)
    most_expensive = find_most_expensive(inventory)
    low_stock_items = get_low_stock_items(inventory)
    print(f"Total Inventory Value: ${total_value:.2f}")
    if most_expensive:
        print(f"Most Expensive Product: {most_expensive['name']} at ${most_expensive['price']:.2f}")
    if low_stock_items:
        print("Low Stock Items:")
        for item in low_stock_items:
            print(f" - {item['name']} (Quantity: {item['quantity']})")
        else:
            print("No low stock items.")
    
    
    print("="*50)

# Main program execution
def main():
    print("Welcome to the Inventory Management System!")
    print("Enter product details (type 'done' for name to finish)")
    
    
    # Get inventory from user
    inventory = add_products()
    
    if not inventory:
        print("No products entered. Exiting...")
        return
    
    # Generate report
    generate_report(inventory)

# Run the program
if __name__ == "__main__":
    main()
```
